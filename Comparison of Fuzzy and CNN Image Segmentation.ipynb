```python
import matplotlib.pyplot as plt
import numpy as np

aspects = ['Learning Capability', 'Accuracy', 'Complexity Handling', 'Generalization', 'Scalability', 'Data Requirements', 'Time to Implementation']
cnn_scores = [9, 9, 9, 9, 10, 5, 4]
fuzzy_scores = [4, 5, 5, 4, 3, 8, 9]
x = np.arange(len(aspects))
width = 0.35

fig, ax = plt.subplots(figsize=(10, 6))
ax.bar(x - width/2, cnn_scores, width, label='CNN')
ax.bar(x + width/2, fuzzy_scores, width, label='Fuzzy Logic')

ax.set_xlabel('Comparison Aspects')
ax.set_ylabel('Performance (1-10)')
ax.set_title('Comparison of CNN vs Fuzzy Logic in Image Segmentation')
ax.set_xticks(x)
ax.set_xticklabels(aspects, rotation=45, ha="right")
ax.legend()
plt.tight_layout()
plt.show()
```
